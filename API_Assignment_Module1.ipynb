# Fetching and Analyzing Country Data Using a Public REST API

```python
import requests

url = "https://restcountries.com/v3.1/all?fields=name,population,region,area"

try:
    response = requests.get(url)
    
    print("Status Code:", response.status_code)
    
    if response.status_code == 200:
        data = response.json()
        print("Total countries retrieved:", len(data))
    else:
        print("Error:", response.text)

except Exception as e:
    print("Error:", e)
```

```python
import pandas as pd

if data is not None:
    country_list = []

    for country in data:
        name = country.get("name", {}).get("common", "Unknown")
        population = country.get("population", 0)
        region = country.get("region", "Unknown")
        area = country.get("area", 0)

        country_list.append({
            "common_name": name,
            "population": population,
            "region": region,
            "area": area
        })

    df = pd.DataFrame(country_list)

    print("\nFirst 10 rows:")
    print(df.head(10))
else:
    print("Skipping processing because data was not fetched.")
```

```python
region_population = df.groupby("region")["population"].sum()

max_region = region_population.idxmax()
max_population = region_population.max()

print("\nRegion with highest population:")
print(f"{max_region} → {max_population}")
```

```python
largest_country = df.loc[df["area"].idxmax()]

print("\nCountry with largest area:")
print(f"{largest_country['common_name']} → {largest_country['area']}")
```
